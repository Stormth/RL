## Part B: NumPy Basics

### 1. Array Creation
- Create a 1D NumPy array with values 0–19.
- Reshape into a 4×5 array.
- Extract the second column and the last row.

````python
import numpy as np  # 导入 NumPy 库并简写为 np,用于数组运算

# 创建一个包含 0~19 共 20 个整数的一维数组
a = np.arange(20)
# 打印这个一维数组
print("1D数组:\n", a)

# 将一维数组重塑为 4 行 5 列的二维数组(4×5)
b = a.reshape(4, 5)
# 打印这个 4×5 二维数组
print("\n4×5数组:\n", b)

# 提取第二列(列索引为 1),得到每一行的第 2 个元素
c = b[:, 1]
# 打印第二列
print("\n第二列:\n", c)

# 提取最后一行(行索引 -1 表示倒数第一行)
d = b[-1, :]
# 打印最后一行
print("\n最后一行:\n", d)
````

### 2. Array Operations
- Generate two random arrays of shape (5,).
- Compute element-wise sum, product, and dot product.
- Normalize one array so its values range between 0 and 1.

````python
import numpy as np  # 导入 NumPy 库并简写为 np

# 生成形状为 (5,) 的随机数组,元素在 [0, 1) 区间
a = np.random.rand(5)
# 再生成一个形状为 (5,) 的随机数组
b = np.random.rand(5)
# 打印第一个随机数组
print("数组1:\n", a)
# 打印第二个随机数组
print("数组2:\n", b)

# 逐元素相加(对应位置求和)
sum_ab = a + b
# 打印逐元素求和的结果
print("\n元素和:\n", sum_ab)

# 逐元素相乘(对应位置求积)
product_ab = a * b
# 打印逐元素相乘的结果
print("\n元素积:\n", product_ab)

# 计算两个数组的点积(内积),即对应元素乘积之和
dot = np.dot(a, b)
# 打印点积结果
print("\n点积:\n", dot)

# 对数组 a 做 Min-Max 归一化,将数值映射到 [0, 1] 区间
normalized_a = (a - np.min(a)) / (np.max(a) - np.min(a))
# 打印归一化后的数组
print("\n归一化后的数组1:\n", normalized_a)
````

### 3. Statistics & Linear Algebra
- Generate a 3×3 random matrix.
- Compute its mean, variance, and determinant.
- Find eigenvalues and eigenvectors.

````python
import numpy as np  # 导入 NumPy 库并简写为 np

# 生成一个 3×3 的随机矩阵,元素在 [0, 1) 区间
a = np.random.rand(3, 3)
# 打印这个随机矩阵
print("随机矩阵:\n", a)

# 计算矩阵所有元素的均值
mean = np.mean(a)
# 计算矩阵所有元素的方差
variance = np.var(a)
# 计算矩阵的行列式
determinant = np.linalg.det(a)
# 打印均值
print("\n矩阵均值:", mean)
# 打印方差
print("矩阵方差:", variance)
# 打印行列式
print("矩阵行列式:", determinant)

# 计算矩阵的特征值和特征向量(特征向量按列排列)
eigenvalues, eigenvectors = np.linalg.eig(a)

# 打印特征值
print("\n特征值:\n", eigenvalues)
# 打印特征向量(每一列对应一个特征向量)
print("\n特征向量:\n", eigenvectors)
````

## Part C: Matplotlib Basics

### 1. Line Plot
- Plot y = sin(x) for x from 0 to 2π.
- Add axis labels and title.

````python
import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
x=np.linspace(0,2*np.pi,100)
plt.plot(x,np.sin(x),label='sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin(x)')
plt.legend()
plt.grid(True)
plt.show()
````

### 2. Multiple Plots
- Plot y = sin(x) and y = cos(x) on the same figure.
- Add legend.

````python
import numpy as np
import matplotlib.pyplot as plt

# 生成从 0 到 2π 的 200 个等间距点
x = np.linspace(0, 2 * np.pi, 200)

# 创建画布和坐标轴
plt.figure(figsize=(8, 4))
# 绘制正弦曲线
plt.plot(x, np.sin(x), label='sin(x)', color='blue')
# 绘制余弦曲线
plt.plot(x, np.cos(x), label='cos(x)', color='red')
# 添加 x 轴标签
plt.xlabel('x')
# 添加 y 轴标签
plt.ylabel('y')
# 添加图表标题
plt.title('sin(x) and cos(x)')
# 显示图例
plt.legend()
# 添加网格线
plt.grid(True)
# 显示图表
plt.show()
````

### 3. Bar Chart
- Simulate rolling a six-sided die 1,000 times.
- Plot a bar chart of frequency of each outcome.

````python
import numpy as np
import matplotlib.pyplot as plt

# 模拟掷一个六面骰子 1000 次,结果范围 1~6
rolls = np.random.randint(1, 7, size=1000)

# 统计每个面出现的次数
outcomes = np.arange(1, 7)
counts = [np.sum(rolls == face) for face in outcomes]

# 创建画布
plt.figure(figsize=(8, 4))
# 绘制柱状图
plt.bar(outcomes, counts, color='steelblue', edgecolor='black')
# 添加 x 轴标签
plt.xlabel('Outcome')
# 添加 y 轴标签
plt.ylabel('Frequency')
# 添加图表标题
plt.title('Frequency of Die Rolls (1000 simulations)')
# 设置 x 轴刻度为 1~6
plt.xticks(outcomes)
# 添加网格线
plt.grid(axis='y', alpha=0.5)
# 显示图表
plt.show()
````

### 4. Histogram
- Generate 1,000 samples from a normal distribution.
- Plot a histogram.
- Overlay a vertical line for the mean.

````python
import numpy as np
import matplotlib.pyplot as plt

# 从标准正态分布中生成 1000 个样本
samples = np.random.randn(1000)
# 计算样本均值
mean_val = np.mean(samples)

# 创建画布
plt.figure(figsize=(8, 4))
# 绘制直方图
plt.hist(samples, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
# 在均值处添加一条垂直红色虚线
plt.axvline(mean_val, color='red', linestyle='dashed', linewidth=2, label=f'Mean = {mean_val:.2f}')
# 添加 x 轴标签
plt.xlabel('Value')
# 添加 y 轴标签
plt.ylabel('Frequency')
# 添加图表标题
plt.title('Histogram of 1000 Normal Samples')
# 显示图例
plt.legend()
# 添加网格线
plt.grid(axis='y', alpha=0.5)
# 显示图表
plt.show()
````

## Part D: Dice Sums
- Simulate rolling two dice 10,000 times.
- Compute probability distribution of sums (2–12).
- Plot distribution as a bar chart.
- Bonus: Compare with theoretical distribution.

````python
import numpy as np
import matplotlib.pyplot as plt

# 模拟掷两个骰子 10000 次
die1 = np.random.randint(1, 7, size=10000)
die2 = np.random.randint(1, 7, size=10000)
# 计算每次的点数之和
sums = die1 + die2

# 统计每个和出现的概率
outcomes = np.arange(2, 13)
empirical_probs = [np.sum(sums == s) / 10000 for s in outcomes]

# 计算理论概率: 两个骰子之和为 s 的组合数 / 36
theoretical_counts = [min(s - 1, 13 - s) for s in outcomes]
theoretical_probs = [c / 36 for c in theoretical_counts]

# 创建画布
plt.figure(figsize=(10, 5))
# 设置柱状图的宽度和位置
width = 0.35
x = np.arange(len(outcomes))
# 绘制模拟概率柱状图
plt.bar(x - width / 2, empirical_probs, width, label='Simulated', color='steelblue', edgecolor='black')
# 绘制理论概率柱状图
plt.bar(x + width / 2, theoretical_probs, width, label='Theoretical', color='orange', edgecolor='black')
# 设置 x 轴刻度
plt.xticks(x, outcomes)
# 添加 x 轴标签
plt.xlabel('Sum of Two Dice')
# 添加 y 轴标签
plt.ylabel('Probability')
# 添加图表标题
plt.title('Probability Distribution of Dice Sums (10,000 simulations)')
# 显示图例
plt.legend()
# 添加网格线
plt.grid(axis='y', alpha=0.5)
# 显示图表
plt.show()
````
